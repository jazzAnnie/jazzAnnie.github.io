## 데이터를 저장하고 읽기

## 파이 그래프는 되도록 쓰지 말자
 * 비교분석이 힘들다.
 * 혹 쓰게 되면 기준을 0시에 잡아서 쓰자: startangle=90

```python
f, ax = plt.subplots(1,2, figsize=(10,5))

not_es_sex['운동을 할 충분한 시간이 없어서'].plot.pie(explode=[0,0.02], ax=ax[0], autopct='%1.1f%%',
                                        startangle=90)
ax[0].set_title('운동을 할 충분한 시간이 없어서')
ax[0].set_ylabel('')

not_es_sex['함께 운동을 할 사람이 없어서'].plot.pie(explode=[0,0.02], ax=ax[1], autopct='%1.1f%%',
                                       startangle=90)
ax[1].set_title('함께 운동을 할 사람이 없어서')
ax[1].set_ylabel('')

plt.show()
```

```python
not_es_age = not_excercise[not_excercise['대분류']=='연령별'].copy()
not_es_age.drop(columns='대분류', inplace=True)
not_es_age.set_index('분류', inplace=True)
not_es_age
```

```python
f, ax = plt.subplots(1,3, figsize=(15,5))
explode_setting = [0.02,0.02,0.02,0.02,0.02,0.02]

not_es_age['운동을 할 충분한 시간이 없어서'].plot.pie(explode=explode_setting, ax=ax[0], autopct='%1.1f%%',
                                        startangle=90)
ax[0].set_title('운동을 할 충분한 시간이 없어서')
ax[0].set_ylabel('')

not_es_age['함께 운동을 할 사람이 없어서'].plot.pie(explode=explode_setting, ax=ax[1], autopct='%1.1f%%',
                                       startangle=90)
ax[1].set_title('함께 운동을 할 사람이 없어서')
ax[1].set_ylabel('')

not_es_age['운동을 할 만한 장소가 없어서'].plot.pie(explode=explode_setting, ax=ax[2], autopct='%1.1f%%',
                                       startangle=90)
ax[2].set_title('운동을 할 만한 장소가 없어서')
ax[2].set_ylabel('')

plt.show()
```
