```python
def create_url(api_node, search_text, start_num, disp_num):
    base = "https://openapi.naver.com/v1/search"
    node = "/" + api_node + ".json"
    param_query = "?query=" + urllib.parse.quote(search_text)
    param_start = "&start=" + str(start_num)
    param_disp = "&display=" + str(disp_num)
    
    return base + node + param_query + param_start + param_disp
```

```python
create_url("book", "TEST", 1, 10)
```

```python
def delete_tag(input_str):
    input_str = input_str.replace("<b>","")
    input_str = input_str.replace("</b>","")
    return input_str
```

```python
def get_dataframe(url):
    client_id = "LP0JME3o423ZaDG2Q87G"
    client_secret = "yEzOfXxjgG"

    request = urllib.request.Request(url)
    request.add_header("X-Naver-Client-Id", client_id)
    request.add_header("X-Naver-Client-Secret", client_secret)

    response = urllib.request.urlopen(request)

    result = json.loads(response.read().decode('utf-8'))

    end_num = result['display']
    
    titles = [delete_tag(result['items'][n]['title']) for n in range(end_num)]
    authors = [delete_tag(result['items'][n]['author']) for n in range(end_num)]
    pubdates = [delete_tag(result['items'][n]['pubdate']) for n in range(end_num)]
    
    return pd.DataFrame({'책제목':titles, '작가':authors, '출판일':pubdates})
```
