## 데이터 프레임의 구조

```python
import pandas as pd

df = pd.read_csv('./auto-mpg.csv', header=None)

df.columns = ['mpg','cylinders','displacement','horsepower','weight','acceleration',
              'model year','origin','name']

print(df.head())
print('\n')
print(df.tail())
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.dtypes)
print('\n')

print(df.mpg.dtypes) # 특정 열의 타입을 알 수 있다.
```

```python
print(df.describe())
print('\n')
print(df.describe(include='all'))
```

```python
#열의 데이터 갯수
print(df.count())
print('\n')

print(type(df.count()))
```

```python
#열의 고유값 갯수
unique_values = df['origin'].value_counts()
print(unique_values)
print('\n')

print(type(unique_values))
```

```python
#통계함수
print(df.mean())
print('\n')
print(df['mpg'].mean())
print(df.mpg.mean())
print('\n')
print(df[['mpg','weight']].mean())
```

```python
# 중간값
print(df.median())
print('\n')
print(df['mpg'].median())
```

```python
#최대값
print(df.max())
print('\n')
print(df['mpg'].max())
```

```python
#최소값
print(df.min())
print('\n')
print(df['mpg'].min())
```

```python
#표준편차
print(df.std())
print('\n')
print(df['mpg'].std())
```

```python
#상관관계
print(df.corr())
print('\n')
print(df[['mpg','weight']].corr())
```
