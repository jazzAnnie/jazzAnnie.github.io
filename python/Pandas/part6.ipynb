# 데이터프레임의 다양한 응용
* **Series객체.apply(매핑함수)** : 시리즈 원소에 함수 매핑 <br>
  인자로 전달하는 매핑 함수에 시리즈의 **모든 원소를 하나씩 입력**하고 함수의 리턴값을 돌려받는다. 

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')

df = titanic.loc[:, ['age','fare']]
df['ten'] = 10
print(df.head())
```

```python
def add_10(n):
    return n+10

def add_two_obj(a, b):
    return a+b

print(add_10(10))
print(add_two_obj(10,10))
```

```python
sr1 = df['age'].apply(add_10) # n=df['age']의 모든 원소
print(sr1.head())
print('\n')

sr2 = df['age'].apply(add_two_obj, b=10) # a=df['age']의 모든 원소,b=10
print(sr2.head())
print('\n')

sr3 = df['age'].apply(lambda x: add_10(x))  # x=df['age']
print(sr3.head())
```

* **데이터프레임 객체.applymap(매핑함수)** : 데이터프레임 원소에 함수 매핑 <br>
  인자로 전달하는 매핑 함수에 시리즈의 **모든 원소를 하나씩 입력**하고 함수의 리턴값을 돌려받는다. 

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')

df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

def add_10(n):
    return n+10

df_map = df.applymap(add_10)
print(df_map.head())
```

## 시리즈 객체에 함수 매핑
* 데이터프레임의 각 열에 함수 매핑
* apply(axis=0) : 모든 열을 하나씩 분리하여 매핑함수의 인자로 각 열(시리즈)가 전달
* DataFrame 객체.apply(매핑함수, axis=0)

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

# 함수는 시리즈를 입력받아서 시리즈를 리턴한다.
def missing_value(series):
    return series.isnull()

result = df.apply(missing_value, axis=0)
print(result.head())
print('\n')
#리턴값은 데이터프레임이다
print(type(result))
```

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

def min_max(x):
    return x.max() - x.min()

result = df.apply(min_max) #df의 각 열에(시리즈) min_max를 적용한다.
print(result)
print('\n')
print(type(result))
```

* apply(axis=1) : 모든 행을 하나씩 분리하여 매핑함수의 인자로 각 행(시리즈)가 전달
* DataFrame 객체.apply(매핑함수, axis=1)

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
df['ten'] = 10
print(df.head())
print('\n')


def add_two_obj(a, b):
    return a+b

df['add'] = df.apply(lambda x: add_two_obj(x['age'], x['ten']), axis=1)
print(df.head())
```

```python
df.describe().T  # describe를 .T로해서 보면 더 쉽게 보인다
```

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

def missing_value(x):
    return x.isnull()

def missing_count(x):
    return missing_value(x).sum()

def total_number_missing(x):
    return missing_count(x).sum()

result_df = df.pipe(missing_value) 
print(result_df.head())
print('\n')
print(type(result_df))
```

```python
result_series = df.pipe(missing_count) 
print(result_series.head())
print('\n')
print(type(result_series))
```

```python
result_value = df.pipe(total_number_missing) 
print(result_value)
print('\n')
print(type(result_value))
```

##  필터링
* 특정 조건식을 만족하는 원소만 따로 추출

* **Boolean indexing** : 참 거짓을 판별하여 (참, 거짓)값으로 구성된 시리즈를 반환, 

 DataFrame 객체[불린 시리즈]

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')

mask1 = (titanic.age >=10) & (titanic.age <20)
df_teenage = titanic.loc[mask1, :]
print(df_teenage.head())
```

```python
mask2 = (titanic.age <10) & (titanic.sex == 'female')
df_female_under10 = titanic.loc[mask2, :]
print(df_female_under10.head())
```

```python
mask3 = (titanic.age <10) | (titanic.age >= 60)
df_under10_morethan60 = titanic.loc[mask3, ['age','sex','alone']]
print(df_under10_morethan60.head())
```

# isin() 메소드
* 특정 값을 가진 행들을 따로 추출

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')

pd.set_option('display.max_columns', 10)  #출력할 열의 갯수를 제한

mask3 = titanic['sibsp'] == 3
mask4 = titanic['sibsp'] == 4
mask5 = titanic['sibsp'] == 5
df_boolean = titanic[mask3 | mask4 | mask5]
print(df_boolean.head())
```

```python
isin_filter = titanic['sibsp'].isin([3,4,5])
df_isin = titanic[isin_filter]
print(df_isin.head())
```

## 데이터 프레임 합치기
* concat() : 데이터를 이어 붙이는 개념
* merge()  : SQL의 join처럼 병합하는 개념
* join()

```python
import pandas as pd
df1 = pd.DataFrame({'a':['a0','a1','a2','a3'],
                   'b':['b0','b1','b2','b3'],
                   'c':['c0','c1','c2','c3']},
                  index=[0,1,2,3])
df2 = pd.DataFrame({'a':['a2','a3','a4','a5'],
                   'b':['b2','b3','b4','b5'],
                   'c':['c2','c3','c4','c5'],
                   'd':['d2','d3','d4','d5']},
                  index=[2,3,4,5])
print(df1, '\n')
print(df2, '\n')

result1 = pd.concat([df1,df2])
print(result1, '\n')
```

```python
result2 = pd.concat([df1,df2], ignore_index=True)
print(result2, '\n')
```

```python
result3 = pd.concat([df1,df2], axis=1)
print(result3, '\n')
```

```python
result3_in = pd.concat([df1,df2], axis=1, join='inner')
print(result3_in, '\n')
```

## 데이터프레임 병합
* **merge()** 는 기준이 되는 열이나 인덱스를 Key라고 부르고 반드시 양쪽 데이터프레임에 모두 존재해야 한다.
* pandas.merge(df_left, df_right, how='inner', on='None')


```python
import pandas as pd

pd.set_option('display.max_columns', 10) # 출력할 최대 열의 개수
pd.set_option('display.max_colwidth', 30)# 출력할 열의 너비
pd.set_option('display.unicode.east_asian_width', True) # 유니코드 사용 너비 조정

df1 = pd.read_excel('./stock price.xlsx')
df2 = pd.read_excel('./stock valuation.xlsx')

print(df1)
print('\n')
print(df2)
```

* **on=None** : 두 데이터 프레임에 **공통으로 속하는 모든 열**을 기준(키)로 병합한다는 뜻
* **how='inner'** : 기준이 되는 **열의 데이터**가 양쪽 데이터프레임에 **공통으로 존재하는 교집합**일 경우 추출한다

```python
# 기본값은 on=None, how='inner
merge_inner = pd.merge(df1, df2)
print(merge_inner)
```

```python
# 합집합
merge_outer = pd.merge(df1, df2, how='outer', on='id')
print(merge_outer)
```

```python
merge_left = pd.merge(df1, df2, how='left', left_on='stock_name', right_on='name')
print(merge_left)
```

```python
merge_right = pd.merge(df1, df2, how='right', left_on='stock_name', right_on='name').sort_values(by='id_x')
print(merge_right)
```

```python
price = df1[df1['price']<50000]
print(price.head())
print('\n')

value = pd.merge(price, df2)
print(value)
```

## 데이터프레임 결합
* join(): merge()함수를 기반으로 만들어졌기 때문에 기본 동작이 서로 비슷
* join 은 두 데이터프레임의 행 인덱스를 기준으로 결합하는 점이 merge와 다르다
* 하지만 on=keys 옵션을 설정하면 행 인덱스 대신에 다른 열을 기준으로 결합할 수 있다.

```python
import pandas as pd

pd.set_option('display.max_columns', 10) # 출력할 최대 열의 개수
pd.set_option('display.max_colwidth', 20)# 출력할 열의 너비
pd.set_option('display.unicode.east_asian_width', True) # 유니코드 사용 너비 조정

df1 = pd.read_excel('./stock price.xlsx', index_col='id')
df2 = pd.read_excel('./stock valuation.xlsx', index_col='id')

df3 = df1.join(df2)
print(df3)
```

```python
df4 = df1.join(df2, how='inner')
print(df4)
```
