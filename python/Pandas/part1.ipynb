## 시리즈
* 시리즈: 순차적으로 나열된 **1차원 배열**의 형태: index와 value관계
* 파이썬의 **딕셔너리와 유사한 구조 {key:value}**
* 따라서 딕셔너리를 시리즈로 변환하는 방법을 많이 사용: pandas.Series(딕셔너리)


```python
import pandas as pd

#딕셔너리 데이터
dict_data = {'a':1, 'b':2, 'c':3}

sr = pd.Series(dict_data)

print(type(sr))
print('\n')
print(sr)
```

```python
# 리스트를 시리즈로 변환하면 디폴트 인덱스가 지정된다.
list_data = ['2019-01-02', 3.14, 'ABC',100, True]
sr = pd.Series(list_data)
print(sr)
```

```python
idx = sr.index
val = sr.values
print(idx)
print('\n')
print(val)
```

```python
#투플을 시리즈로 변환
tup_data = ('영인','2010-05-01','여', True)
sr = pd.Series(tup_data, index=['이름', '생년월일','성별','학생여부'])
print(sr)
```

```python
print(sr[0]) #정수형 위치 인덱스: 1번째 원소를 선택
print(sr['이름']) #인덱스 이름을 이용한 참조
```

```python
print(sr[[1,2]])
print('\n')
print(sr[['생년월일','성별']])
```

```python
print(sr[1:2]) #2는 포함되지 않는다
print('\n')
print(sr['생년월일':'성별']) # 성별은 포함된다.
```

```python
# NaN값이 있는 시리즈 연산
import pandas as pd
import numpy as np

student1 = pd.Series({'국어':np.nan, '영어':80, '수학':90})
student2 = pd.Series({'수학':80, '국어':90})
print(student1)
print('\n')
print(student2)
print('\n')

addition = student1 + student2
subtraction = student1 - student2
multiplication = student1 * student2
division = student1 / student2
print(type(division))
print('\n')

#addition등의 key값이 컬럼으로 바뀐다
result = pd.DataFrame([addition, subtraction, multiplication, division],
                     index=['덧셈','뺄셈','곱셈','나눗셈'])
print(result)
```

```python
# NaN상황을 피하기 위해서는 NaN 대신 0을 입려할 수 있다.
student1 = pd.Series({'국어':np.nan, '영어':80, '수학':90})
student2 = pd.Series({'수학':80, '국어':90})
print(student1)
print('\n')
print(student2)
print('\n')

sr_add = student1.add(student2, fill_value=0)
sr_sub = student1.sub(student2, fill_value=0)
sr_mul = student1.mul(student2, fill_value=0)
sr_div = student1.div(student2, fill_value=0)

#addition등의 key값이 컬럼으로 바뀐다
result = pd.DataFrame([sr_add, sr_sub, sr_mul, sr_div],
                     index=['덧셈','뺄셈','곱셈','나눗셈'])
print(result)
```

# 웹 스크래핑

```python
from bs4 import BeautifulSoup
import requests
import re
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_American_exchange-traded_funds"
resp = requests.get(url)
soup = BeautifulSoup(resp.text, 'lxml') #lxml: 기본적으로 제공하는 파서
rows = soup.select('div > ul > li') #div를 부모로, ul을 부모로, li를 요소로 (CSS스타일 뽑아오기)

print(rows[0])
```

```python
etfs = {}
for row in rows:
    try:
        etf_name = re.findall('^(.*) \(NYSE', row.text) # (.*) 값을 etf_name에 넣는다
        etf_market = re.findall('\((.*)\|', row.text)
        etf_ticker = re.findall('NYSE Arca\|(.*)\)', row.text)
        
        if(len(etf_ticker) > 0) & (len(etf_market) > 0):
            if((len(etf_name)) > 0): # 이 구문이 없으면 index out of range에러가 난다.
                etfs[etf_ticker[0]] = [etf_market[0], etf_name[0]]

    except AttributeError as err:
        pass

print(etfs)
print('\n')

df = pd.DataFrame(etfs)
print(df)
```
