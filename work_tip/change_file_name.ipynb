## 동일 패턴인 파일명 일괄 수정

- (조건) 동일한 디렉토리안에 AAAAA파일1.xlsx, AAAAA파일2.xlsx, ...AAAAA라는 이름이 계속 반복되는 여러개 파일존재
- AAAAA라는 이름을 BBBBB라고 바꾸고 싶음

```python
import os

file_path = './../../test'   # 바꾸고자 하는 파일 목록이 여기에 들어 있다, 한글이 안됨
file_names = os.listdir(file_path)
file_names
```

```python
for file in file_names:
    if file[-2:] != 'jpg':
        name = file
        new_name = file.replace('진달래', '진달래00')
        name = os.path.join(file_path, name)
        new_name = os.path.join(file_path, new_name)
        os.rename(name, new_name)
```
