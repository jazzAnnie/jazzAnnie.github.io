```python
# 필요한 컬럼만 지정해서 가져온다.
df_sheet_index = pd.read_excel('./../../test/0. 전산실현황(서울특별시교육청).xlsx', sheet_name=1
                               ,skiprows=[0,1,2], usecols="B:BI")
print(df_sheet_index)
```

```python
# 행열 바꾸기
df_tx = df_sheet_index.transpose()
df_tx.head()
```

```python
# 컬럼지정하기
from pandas import DataFrame
df_col= DataFrame(df_tx)
df_col=df_col.rename(columns={0: '구분1', 1: '구분2', 2:'교수학습정보부', 3:'종합전산센터', 4:'재해복구센터'})
df_col[48:55]
```

```python
# 결측치 채우기 (50번~54까지가 NaN으로 되어 있음) - 앞의 값으로 채움
import numpy as np 

df_col['구분1'].fillna(method='ffill', inplace=True)
df_col[48:55]
```

```python
# 구분2의 NaN 를 스페이스로 변경하기
df_col['구분2'].replace(np.NaN, '', inplace=True)
df_col[54:55]
```

```python
# 컬럼의 타입을 string 으로 각각 변경한다.
# df_col[['구분1', '구분2']].astype(str)
df_col[19:28]
```

```python
# 컬럼 값을 합치기
df_col['구분']=df_col['구분1'].str.cat(df_col['구분2'].astype(str))
df_col[19:28]
```

```python
# 인덱스를 지정한다.
df_col.set_index(df_col.구분, inplace=True)
df_col[19:28]
```

```python
# 필요없는 컬럼을 지운다. axis row:0  column:1
df_col=df_col.drop(['구분1','구분2','구분'], axis=1) 
df_col[19:28]
```
